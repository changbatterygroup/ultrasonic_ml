## Loading and analyzing a single file <a name="load"></a>

````python
# First we load the data from the file we want to analyze
# The experiment should always output sqlite3 files, so let's convert them to a more usable pickle form first
sqliteFile = "/home/xinqiao/new_mount/gaussian_sampler/ultrasound_data/fatima_pouch_cell_transmission_RT_to_40C.sqlite3"

# Convert the sqlite3 to a .pickle
# This takes a few seconds. A progress bar will display in your terminal
pj.sqliteToPickle(sqliteFile)

# Load the pickle
pickleFile = os.path.splitext(sqliteFile)[0] + '.pickle'
data = pj.loadPickle(pickleFile)
````

Your data is now loaded as a dict of dicts. Some notes on the structure of the data:

Top level keys are the collection_index of each waveform. Values are dicts whose keys are the data stored for each collection_index

In addition, there are two top level keys `'fileName' = pickleFile` and `'parameters'` which contain a dict of the pulse parameters

```
data = {     
        0 : {'voltage' : np.array(...), 'time' : np.array(...), 'time_collected' :, ... },
        1 : {'voltage' : np.array(...), 'time' : np.array(...), 'time_collected' :,... },
          ...
        'parameters' : { 'measure_time' : .., 'delay' : .. ,...},
        'fileName' : pickleFile 
        }
```

Now let's do some basic analysis.

`applyFunctionToData` takes a user-input analysis function and applies it to all your data

It takes at least 4 arguments:
- `dataDict` - the data object created by using `pj.loadPickle()`
-  `func` - the function to be applied to the data
-  `resKey` - the key the result will be stored in within dataDict
-  `dataKeys` - the data that will be input into func, as an ordered list. For example `['voltage', 'time']`

Here's a simple example which calculates the maximum voltage of each wave within data and saves it as `'max'`


````python
data = pj.applyFunctionToData(data, np.max, 'max', ['voltage'])
````

`applyFunctionToData` can also take additional arguments if `func` requires them. 
For example, the function `pj.staltaFirstBreak` calculates the first break time of a wave by the STA/LTA (Short-Term Average/Long-Term Average) algorithm. 

This function requires several user defined paramaters in addition to the voltage and time: the short window (the window to perform short time averaging), the long window (for long time averaging) and the threshold (a number in (0,1) ) to determine what counts as the wave breaking

In the example below, we apply the STA/LTA algorithm using a short window of 5, a long window of 30, and a threshold of 0.75
- NOTE: the order that dataKeys and extra *funcargs are input matters! They must follow the input specified where the function is defined
- NOTE: the extra parameters are not saved anywhere. It is good practice to include them in the resKey for record keeping

````python
data = pj.applyFunctionToData(data, pj.staltaFirstBreak, 'STA/LTA_5_30_0d75', ['voltage', 'time'], 5, 30, 0.75)
````

Here's a more complicated example. Let's calculate the real parts of the fft of the waveform
we can do this using the numpy function np.fft.rfft()


````python
data = pj.applyFunctionToData(data, np.fft.rfft, 'fft', ['voltage'])
````

````python
# To make a nice plot, we want the magnitude of the rfft by applying abs() to the result
data = pj.applyFunctionToData(data, abs, 'abs_fft', ['fft'])
````

````python
# Next we calculate the fft frequencies using np.fft.rfftfreq(). This is handled a little differently since it is a constant array for all scans
# rfftfreq requires two inputs: the number of samples (= len('time') ), and the time step
# Since every scan should have the same 'time' values, we only need to calculate this once and set it for all data
timeStep = (data[0]['time'][1] - data[0]['time'][0]) / 1000000000
numSamples = len(data[0]['time'])
fft_freq = np.fft.rfftfreq(numSamples, timeStep)
````

````python
# the function pj.writeDataToDict can be used to write a constant value for every collection_index
data = pj.writeDataToDict(data, fft_freq, 'fft_freq')
````

Now we have the fft and its corresponding frequencies saved in the data under the keys 'abs_fft' and 'fft_freq'
This data is also saved - any time we load the pickle in the future, it will be accessible

There are several helper functions written in pickleJar.py for analysis in this format. A list is provided at the end
with an explanation of the calculation and what is needed for their input

## Plotting a scan <a name="plot_scan"></a>

````python
pickleFile = "/home/xinqiao/new_mount/gaussian_sampler/ultrasound_data/fatima_pouch_cell_transmission_RT_to_40C.pickle"
data = pj.loadPickle(pickleFile)
````

````python
import plotly.graph_objects as go
from ipywidgets import interact
import ipywidgets as widgets
from datetime import datetime
import matplotlib.cm as cm

def plot_waveform(i):
    fig = go.Figure()
    fig.add_trace(go.Scatter(x=data[i]['time'], y=data[i]['voltage'], mode='lines'))
    fig.update_layout(
        title=f'Measurement {i} at ({datetime.fromtimestamp(data[i]["time_collected"]).strftime("%Y-%m-%d, %H:%M:%S")})',
        xaxis_title='Time (ns)',
        yaxis_title='Voltage (mV)'
    )
    fig.show()

slider = widgets.IntSlider(
    value=0,
    min=0,
    max=len(data) - 2,
    step=1,
    description='measurement:',
    continuous_update=False
)

interact(plot_waveform, i=slider)
````

````python
def plot_waveforms(inds, slice_time=[0,-1], title=''):
    fig = go.Figure()
    # Get viridis colormap
    viridis = cm.get_cmap('viridis')
    n_traces = len(inds)
    for idx, i in enumerate(inds):
        # Map index from 0 to 1 for viridis (darkest to lightest)
        color_val = idx / max(1, n_traces - 1) if n_traces > 1 else 0
        # Get RGB color from viridis (returns RGBA, we only need RGB)
        rgb_color = viridis(color_val)[:3]
        # Convert to plotly color format (rgb(r, g, b))
        plotly_color = f'rgb({int(rgb_color[0]*255)}, {int(rgb_color[1]*255)}, {int(rgb_color[2]*255)})'
        fig.add_trace(go.Scatter(
            x=data[i]['time'][slice_time[0]:slice_time[1]], 
            y=data[i]['voltage'][slice_time[0]:slice_time[1]], 
            mode='lines', 
            name=f'{i} at ({datetime.fromtimestamp(data[i]["time_collected"]).strftime("%Y-%m-%d, %H:%M:%S")})',
            line=dict(width=1, color=plotly_color)
        ))
    fig.update_layout(
        title=title,
        xaxis_title='Time (ns)',
        yaxis_title='Voltage (mV)'
    )
    fig.show()
````

````python
# 1 hr, 5 min intervals
inds = np.arange(0,61,5)
# slice_time = [0, 6000]
# plot_waveforms(inds, slice_time)
plot_waveforms(inds, title='0-1 hr, every 5 min')
````

````python
# 1hr to 6hr, 30 min intervals
inds = np.arange(60,60*6+1,30)
plot_waveforms(inds, title='1-6 hr, every 30 min')
````

````python
# 6hr to 15hr, 60 min intervals
inds = np.arange(60*6,60*15+1,60)
plot_waveforms(inds, title='6-15 hr, every 1 hr')
````

````python
# 15-24 hr, 1 hr intervals
inds = np.arange(60*15+1,len(data)-2,60)
plot_waveforms(inds, title='15-24 hr, every 1 hr')
````

## Analysis functions and usage<a name="appx"></a>

---

### `np.max`
- Returns the **maximum value** of an array
- **Input:** array | **Output:** single value
```python
pj.applyFunctionToData(data, np.max, 'max', ['voltage'])
```

---

### `bn.nanmax`
- Returns the **maximum value** of an array using the `bottleneck` module
- Faster than `np.max`
- **Input:** array | **Output:** single value
```python
pj.applyFunctionToData(data, bn.nanmax, 'max', ['voltage'])
```

---

### `pj.maxMinusMin`
- Returns **max minus min** of an array
- Useful as a measure of signal intensity when the baseline is drifting
- **Input:** array | **Output:** single value
```python
pj.applyFunctionToData(data, pj.maxMinusMin, 'maxMinusMin', ['voltage'])
```

---

### `pj.staltaFirstBreak`
- Returns the **first break time** of a waveform using the STA/LTA algorithm (short term average / long term average)
- **Inputs:** voltage array, time array, plus auxiliary parameters:
  - `short window` — length in time steps (not time)
  - `long window` — length in time steps (not time)
  - `threshold` — fraction of maximum value in range `(0, 1)` that indicates signal arrival
- **Output:** single value
- Since auxiliary parameter values aren't saved, it's good practice to include them in the result key name
```python
# short window = 5, long window = 30, threshold = 0.75
pj.applyFunctionToData(data, pj.staltaFirstBreak, 'staltaFirstBreak_5_30_0d75', ['voltage', 'time'], 5, 30, 0.75)
```

---

### `pj.absoluteSum`
- Returns the **sum of absolute values** of an array
- Directly proportional to the integral of the signal — a useful and fast metric for signal intensity
- **Input:** voltage array | **Output:** single value
```python
pj.applyFunctionToData(data, pj.absoluteSum, 'absoluteSum', ['voltage'])
```

---

### `pj.savgolFilter`
- Applies a **Savitzky-Golay filter** to smooth data and optionally calculate derivatives
- Useful for smoothing noisy data and finding extrema via derivatives
- **Inputs:** y-data array (e.g. `voltage`), x-data array, plus optional auxiliary parameters:
  - `window length` — number of points per fit; higher values give smoother signals but may distort the data
  - `polynomial order` — order used for fitting (default: `3`)
  - `derivOrder` — order of derivative to calculate; must be less than polynomial order (`0` = smoothing only)
- **Output:** array of the same shape as y-data
```python
# Example 1: smooth signal — window=11, polyOrder=3, derivOrder=0
pj.applyFunctionToData(data, pj.savgolFilter, 'savgol_11_3_0', ['voltage', 'time'], 11, 3, 0)

# Example 2: first derivative — window=31, polyOrder=5, derivOrder=1
pj.applyFunctionToData(data, pj.savgolFilter, 'savgol_31_5_1', ['voltage', 'time'], 31, 5, 1)
```

---

### `pj.zeroCrossings`
- Returns the **x-coordinates where y-values cross zero** (change sign)
- Can be combined with first or second derivatives to find extrema or inflection points
- **Inputs:** y-data array, x-data array, plus optional auxiliary parameter:
  - `linearInterp` *(bool, default `False`)* — if `True`, returns x-values from linear interpolation around the sign change; if `False`, returns the x-value just before the sign change
- **Output:** array of x-coordinates
```python
# Find extrema using the first derivative calculated via pj.savgolFilter
pj.applyFunctionToData(data, pj.zeroCrossings, 'extrema', ['savgol_31_5_1', 'time'], True)
```

---

### `pj.listExtrema`
- Returns a list of **(x, y) coordinates of the extrema** of an input function
- **Inputs:** y-data array, derivative array, x-data array
- **Output:** list of `(x, y)` coordinate pairs
```python
# Find extrema using the Savitzky-Golay derivative from a prior step
pj.applyFunctionToData(data, pj.listExtrema, 'extrema_coors', ['voltage', 'savgol_31_5_1', 'time'])
```
